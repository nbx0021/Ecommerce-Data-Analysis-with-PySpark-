```python
import main_pipeline
from pyspark.sql.functions import col

# 1. Run the Pipeline to load the "In-Memory" Tables
# This ensures the views (gold_*) exist in this session
main_pipeline.run_pipeline()

print("--- 🚀 EXECUTIVE DASHBOARD ---")

# ---------------------------------------------------------
# VISUAL 1: Logistics & Delivery Performance
# Best Visual: "Counter" or "Big Number"
# ---------------------------------------------------------
print("1. Logistics KPIs")
df_logistics = spark.sql("""
    SELECT 
        avg_delivery_days, 
        avg_estimated_days, 
        avg_delay_vs_estimate 
    FROM gold_logistics_performance
""")
display(df_logistics)
# 🎨 Plot Setup:
# - Visualization Type: Counter
# - Value: avg_delivery_days
# - Target: avg_estimated_days

# ---------------------------------------------------------
# VISUAL 2: Customer Segments (RFM)
# Best Visual: Pie Chart or Bar Chart
# ---------------------------------------------------------
print("2. Customer Segments (RFM Analysis)")
df_rfm = spark.sql("""
    SELECT * FROM gold_customer_segments 
    ORDER BY avg_spend DESC
""")
display(df_rfm)
# 🎨 Plot Setup:
# - Visualization Type: Pie Chart
# - Values: customer_count
# - Group By: customer_segment

# ---------------------------------------------------------
# VISUAL 3: Monthly Revenue Trend
# Best Visual: Line Chart or Bar Chart
# ---------------------------------------------------------
print("3. Monthly Revenue Trend")
# Removing 'LIMIT 3' so we can see the full trend line
df_trend = spark.sql("""
    SELECT * FROM gold_monthly_sales 
    ORDER BY month
""")
display(df_trend)
# 🎨 Plot Setup:
# - Visualization Type: Line Chart
# - X Column: month
# - Y Column: total_revenue

# ---------------------------------------------------------
# VISUAL 4: Top Products by Revenue
# Best Visual: Horizontal Bar Chart
# ---------------------------------------------------------
print("4. Top Performing Categories")
# Showing Top 10 instead of 3 for a better chart
df_products = spark.sql("""
    SELECT * FROM gold_product_performance 
    LIMIT 10
""")
display(df_products)
# 🎨 Plot Setup:
# - Visualization Type: Bar Chart
# - X Column: total_revenue
# - Y Column: category_name
# - Orientation: Horizontal
```
